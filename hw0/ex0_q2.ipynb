```python
import numpy as np
import matplotlib.pyplot as plt
import torch
```

```python
# 2.1
def f1(x):
  return 4*x -11

samples_x = np.linspace(0, 10, 1000)
samples_y = [f1(x) for x in samples_x]
plt.plot(samples_x,samples_y)
```

```python
# 2.2
def f2(x):
  return x**2 +2*x -1

samples_x = np.linspace(0, 10, 10000)
samples_y = np.array([f2(x) for x in samples_x])
plt.plot(samples_x,samples_y)
```

```python
# 2.3 add sandard gaussian
std_gaussian = np.random.normal(0, 1, 10000)
samples_y += std_gaussian
plt.plot(samples_x,samples_y)
```

```python
# 2.6
import math
def sigmoid(x):
  return math.exp(x) / (1 + math.exp(x))
def sigmoid_derivative(x):
  return math.exp(x) / ((1 + math.exp(x))**2)
samples_x = np.linspace(-10, 10, 1000)
samples_y1 = np.array([sigmoid(x) for x in samples_x])
samples_y2 = np.array([sigmoid(-x) for x in samples_x])
samples_y3 = np.array([sigmoid_derivative(x) for x in samples_x])

plt.plot(samples_x,samples_y1)
plt.plot(samples_x,samples_y2)
plt.plot(samples_x,samples_y3)
plt.show()

```

**results:**
- sigmoid(x) in blue
- sigmoid(-1) in orange
- sigmoid'(x) in green

We can see that sigmoid(-x) is a reflection on the x axis of sigmoid(x)
